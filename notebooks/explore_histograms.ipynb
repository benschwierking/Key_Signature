# GiantSteps chroma histograms by key
This notebook reads the derived 36-feature cache only; it never loads or stores audio.

```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

feature_file = next(Path('../data/features').glob('*.parquet'))
features = pd.read_parquet(feature_file)
histogram_columns = [f'hist_{i}' for i in range(12)]
average = features.groupby('key')[histogram_columns].mean().sort_index()
average.head()
```

```python
fig, axes = plt.subplots(6, 4, figsize=(14, 15), sharey=True)
for ax, (key, values) in zip(axes.flat, average.iterrows()):
    ax.bar(range(12), values)
    ax.set_title(key)
    ax.set_xticks(range(12), ['C','C#','D','D#','E','F','F#','G','G#','A','A#','B'], rotation=90)
fig.suptitle('Average track-wide chroma histogram by annotated key')
fig.tight_layout()
```
